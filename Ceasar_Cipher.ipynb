```python
def caesar_cipher(text, shift):
    result = ""
    for char in text:
        if char.isalpha():
            result += chr((ord(char) - 65 + shift) % 26 + 65)
        else:
            result += char
    return result

text = "AEROXKAXO"
shift = 3
encrypted = caesar_cipher(text, shift)

print(f"Original: {text}")
print(f"Encrypted: {encrypted}")
```
